```python
import sqlite3

conn = sqlite3.connect(r"c:\Users\KRISH JHA\OneDrive\Desktop\Sepsis\data_handling\sepsis.db")

print("Database created successfully!")

```

```python
cursor=conn.cursor()


cursor.execute("""
              CREATE TABLE patient_data(
                Patient_id Text,
              HR Integer,
              O2Sat Integer,
              Temp Integer,
              SBP Integer,
              MAP Integer,
              DBP Integer,
              Resp Integer,
              EtCO2 Integer,
              BaseExcess Integer,
              HCO3 Integer,
              FiO2 Integer,
              pH Integer,
              PaCO2 Integer,
              SaO2 Integer,
              AST Integer,
              BUN Integer,
              Alkalinephos Integer,
              Calcium Integer,
              Chloride Integer,
              Creatinine Integer,
              Bilirubin_direct Integer,
              Glucose Integer,
              Lactate Integer,
              Magnesium Integer,
              Phosphate Integer,
              Potassium Integer,
              Bilirubin_total Integer,
              TroponinI Integer,
              Hct Integer,
              Hgb Integer,
              PTT Integer,
              WBC Integer,
              Fibrinogen Integer,
              Platelets Integer,
              Age Integer,
              Gender Integer,
              Unit1 Integer,
              Unit2 Integer,
              HospAdmTime Integer,
              ICULOS Integer,
              SepsisLabel Integer
              )
              
""")

conn.commit()
```

```python
import pandas as pd
import glob
import os


files=glob.glob(r"C:\Users\KRISH JHA\OneDrive\Desktop\Sepsis\data_handling\training\**\*.psv")
print(len(files))
c=0
for file in files:
    patient_id=os.path.splitext(os.path.basename(file))[0]
    df=pd.read_csv(file,sep='|')
    df['Patient_id']=patient_id
    df.to_sql(
        "patient_data",
        conn,
        if_exists="append",
        index=False
    )
    c+=1
    if (c%1000==0):
        print(f"file{c+1} :Done")

print("All File Done")
```
